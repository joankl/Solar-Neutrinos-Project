# Notebook dedicated to observe the MC data of solar neutrinos on 2.2PPO

```python
import numpy as np
import glob

import matplotlib.pyplot as plt
import seaborn as sn
```

# Load the Data

```python
main_dir = '/home/joankl/data/solars/mc/2p2PPO/'

energy_b8_nue = np.load(main_dir + 'b8Nue/unosc/np_array_resume/energy_corrected.npy')
energy_tl208 = np.load(main_dir + 'tl208/np_array_resume/energy_corrected.npy')

itr = np.load(main_dir + 'b8Nue/unosc/np_array_resume/itr.npy')
```

# Plots

## Energy Spectrum

```python
bins = 100
sn.histplot(energy_b8_nue, bins = bins, element = 'step', alpha = 0, linewidth = 1.2, label = '8B-nue')
sn.histplot(energy_tl208, bins = bins, element = 'step', alpha = 0, linewidth = 1.2, label = '208Tl')

plt.xlabel('Reconstructed Energy (MeV)')
plt.legend()

plt.yscale('log')

plt.title('Energy Spectrum of MC 8B-nue and 208Tl in 2.2PPO')

plt.xlim(2.5,)
plt.show()
```

## ITR

```python
bins = 100
sn.histplot(itr, bins = bins, element = 'step', alpha = 0, linewidth = 1.2)

plt.xlabel('ITR')

plt.yscale('log')

plt.title('ITR of MC 8B-nue in 2.2PPO')

#plt.xlim(0,)
plt.show()
```
